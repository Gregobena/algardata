```python


# Suponiendo que tus datos los cargás en un DataFrame llamado 'df'
# df = pd.read_csv('tu_archivo.csv') o pd.read_excel('tu_archivo.xlsx')

# Agrupar por la columna 'Asignado a'
analisis = df_casos.groupby('Asignado a').agg(
    total_duracion=('Duración', 'sum'),
    cantidad_casos=('Asignado a', 'count')
).reset_index()

# Ordenar para ver quién tuvo MÁS duración en total
top_duracion = analisis.sort_values(by='total_duracion', ascending=False)
print("--- Ranking por mayor duración ---")
print(top_duracion)

# Ordenar para ver quién tuvo MÁS casos asignados
top_casos = analisis.sort_values(by='cantidad_casos', ascending=False)
print("\n--- Ranking por mayor cantidad de casos ---")
print(top_casos)
```
